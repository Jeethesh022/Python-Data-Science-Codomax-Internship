```python
import pandas as pd
import numpy as np

data = pd.read_csv("/content/Datasets.csv")

print(data)
```

```python
data.shape
data.head()
data.tail()
data.columns
data.info()
```

```python
data.describe()
```

```python
Movie=data["Movie"]
Genres=data["Genre"]
Rating=data["Rating"]
year=data["Year"]
Budget=data["Budget"]
```

```python
print("Rating columns:\n")
print(Rating)
```

```python
print("Rating Greater than 8 or not:\n")
print(Rating>8)
```

```python
print("Datas with rating greater than 8:\n")
print(data[Rating>8])
```

```python
print("Sorted based on rating:\n")
print(data.sort_values("Rating"))
```

```python
print("In Decending:\n")
print(data.sort_values("Rating",ascending=False))
```

```python
print("Groupby year with avg rating:\n")
data.groupby("Year")["Rating"].mean()
```

```python
print("with genre:\n")
data.groupby("Genre")["Rating"].mean()
```

```python
print("counts per year:\n")
data["Year"].value_counts()
```

```python
print("Check miss values:\n")
data.isna()
```

```python
print("using loc to identify the movie:\n")
data.loc[17,"Movie"]
```

```python
print("show the list of datas from 0 to 5:\n")
data.iloc[0:6]
```

```python
print("Highet Rating Index of movie:",data["Rating"].idxmax())
```

```python

print("Lowest Rating movie:",data.loc[data["Rating"].idxmin(), "Movie"])
```

```python
print("Highest Rating movie:",data.loc[data["Rating"].idxmax(), "Movie"])
```

```python
print("Highest Budget:",data.loc[data["Budget"].idxmax(), "Movie"])
```

```python
print("Lowest budget:",data.loc[data["Budget"].idxmin(), "Movie"])
```

```python
print("Oldest movie:",data.loc[data["Year"].idxmin(),"Movie"])
```

```python
print("Newest movie:",data.loc[data["Year"].idxmax(),"Movie"])
```

```python
print("Total Number of Movies:",data["Movie"].count())
```

```python
ratings = np.array([8.8, 8.6, 8.4, 8.0, 7.9])
print("Array like ratings:\n")
print(ratings)
```

```python
print("Basic operation:\n")
print(ratings + 1)
print(ratings - 1)
print(ratings * 2)
print(ratings / 2)
```

```python
print("Imp functions:\n")
print("Highest Rating:", np.max(ratings))
print("Lowest Rating:", np.min(ratings))
print("Average Rating:", np.mean(ratings))
print("Total Movies:", np.size(ratings))
```

```python
print("Indexing:\n")
print(ratings[0])
print(ratings[2])
```

```python
print("Two Dimensions Array:\n")
movie_data = np.array([
    [2010, 8.8],
    [2014, 8.6],
    [2009, 8.4],
    [2010, 8.0]
])

print(movie_data)
print(movie_data.shape)
print(movie_data.ndim)
```
