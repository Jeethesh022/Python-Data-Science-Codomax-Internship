# Data Visualization Using Python

In this module, I learned how to visualize movie data using Python libraries such as Matplotlib and Seaborn.

I practiced different charts to understand movie counts, trends, distributions, and relationships between columns.

## Libraries Used

I used the following Python libraries for data visualization:

- Pandas: To read and analyze the movie dataset.
- Matplotlib: To create basic charts.
- Seaborn: To create advanced and attractive charts.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sea
```

## Loading the Dataset

I loaded the Hollywood movies dataset using Pandas.

The dataset contains information about:

- Movie name
- Genre
- Year
- Rating
- Budget

This data was used to create different visualizations.

```python
data=pd.read_csv("/content/hollywood_movies_100.csv")
data.shape
data.info()
data.head()
```

## Bar Chart

A bar chart is used to compare values between different categories.

I used a bar chart to compare the number of movies in each genre,Average rating per each genre and year wise movies count

It helps me understand which genres have more movies and which have fewer movies.

```python
data["Genre"].unique()
```

```python
genre_count=data.groupby("Genre").size()
```

```python
print(genre_count)
```

```python
plt.bar(genre_count.index,genre_count.values)
plt.title("Movies count per genre")
plt.xlabel("Genre list")
plt.ylabel("Movie count")
plt.show()
```

```python
print("Max movies at one genre")
genre_count.idxmax()
```

```python
genre_rating=data.groupby("Genre")["Rating"].mean()
print(genre_rating)
```

```python
plt.bar(genre_rating.index,genre_rating.values)
plt.title("Avg rating in particular genre")
plt.xlabel("Genre list")
plt.ylabel("Avg rating")
plt.show()
```

```python
print("highest avg rating in one genre")
genre_rating.idxmax()
```

```python
year_count=data.groupby("Year").size()
print(year_count)
```

```python
plt.bar(year_count.index,year_count.values,color="red")
plt.title("Movie released per year")
plt.xlabel("Released year")
plt.ylabel("Movies count")
plt.show()
```

```python
print("Highest movies released in the one year\n")
year_count.idxmax()
```

```python
genre_data = genre_count.reset_index()
print(genre_data)
```

```python
sea.barplot(data=genre_data, x="Genre", y=0)
```

## Line Chart

A line chart is used to show trends and changes over time.

I used a line chart to show the number of movies released in each year.

It helps me understand whether the movie count is increasing, decreasing, or remaining the same.

```python
from matplotlib.typing import MarkerType
plt.plot(year_count.index,year_count.values,marker="o",linestyle='-')
plt.title("Films per year")
plt.xlabel("Year")
plt.ylabel("Films")
plt.show()
```

## Pie Chart

A pie chart is used to show the percentage or proportion of each category in a total.

I used a pie chart to show the percentage of movies in each genre.

It helps me understand how much each genre contributes to the total number of movies.

```python
plt.pie(
    genre_count.values,
    labels=genre_count.index,
    autopct="%1.1f%%"
)

plt.title("Genre-wise Movie Distribution")
plt.show()
```

## Scatter Plot

A scatter plot is used to understand the relationship between two numerical columns.

I used a scatter plot to compare movie budget and rating.

Each dot represents one movie.

The graph showed a very weak positive relationship between budget and rating.

```python
plt.scatter(data["Budget"], data["Rating"])

plt.title("Budget vs Rating")
plt.xlabel("Budget")
plt.ylabel("Rating")

plt.show()
```

```python
data["Budget"].corr(data["Rating"])
```

## Histogram

A histogram is used to understand the distribution of numerical values.

I used a histogram to show how movie ratings are distributed across different ranges.

It helps me identify which rating range contains the most movies.

```python
plt.hist(data["Rating"], bins=10)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")

plt.show()
```

## Visualization Using Seaborn

I also used Seaborn to create visualizations.

Seaborn provides a simple way to create attractive statistical charts.

I practiced a bar chart to compare movie counts by genre and a box plot to compare ratings across genres.

```python
genre_count = data.groupby("Genre").size()

genre_data = genre_count.reset_index()
genre_data.columns = ["Genre", "Movie_Count"]

sea.barplot(data=genre_data, x="Genre", y="Movie_Count")

plt.title("Genre-wise Movie Count")
plt.xlabel("Genre")
plt.ylabel("Movie Count")

plt.show()
```
