Hollywood Movie Data Example

1. Variables
Store basic information about a movie.

```python
movie_name = "Inception"
director = "Christopher Nolan"
year = 2010
rating = 8.8
budget = 160
print(movie_name)
print(director)
print(year)
print(rating)
print(budget)
```

List
Store multiple movie names.

```python
movies = ["Inception", "Interstellar", "Avatar", "Titanic", "Avengers"]

print(movies)
```

```python
movies[3]
```

Dictionary
Store complete details of a movie.

```python
movie = {
    "name": "Inception",
    "director": "Christopher Nolan",
    "year": 2010,
    "rating": 8.8,
    "budget": 160
}

print(movie)
```

2.Data Types

Python supports different types of data such as:

String (str) → Movie name, Director
Integer (int) → Year, Budget
Float (float) → Rating


```python
print(type(movie["name"]))
print(type(movie["director"]))
print(type(movie["year"]))
print(type(movie["rating"]))
```

3. Input and Output
input() is used to get data from the user, and print() is used to display output.

```python
print("Year of movie", movie["year"])
print("Rating", movie["rating"])

Movie = input("Enter ur fav movie: ")
print(Movie)
```

4.Operators

Python operators are used to perform calculations and compare values.

```python
print(movie["budget"]+40)#arithemetic
print(movie["budget"]*2)
print(movie["rating"]>8)#comparision
print(movie["year"] == 2010)
print(movie["rating"] > 8 and movie["year"] > 2000)#logical
```

5.Conditional Statements

Conditional statements are used to make decisions based on conditions.

if → checks the first condition,
elif → checks another condition,
else → runs when all conditions are false

```python
rating = movie["rating"]

if rating >= 8.5:
    print("Excellent")
elif rating >= 7.0:
    print("Good")
else:
    print("Average")
```

6.Loop

For Loop
A for loop is used to repeat a task for each item in a collection.

```python
for m in movies:
  print(m)
```

While Loop

A while loop is used to repeat a task as long as a condition is true.

```python
  i = 0

while i < len(movies):
    print(movies[i])
    i = i + 1
```

8.Functions

A function is a reusable block of code used to perform a specific task.

```python
def check_rating(rating):
    if rating >= 8.5:
        print("Excellent")
    elif rating >= 7:
        print("Good")
    else:
        print("Average")

check_rating(movie["rating"])
```

```python
def get_rating(rating):
    return rating
result = get_rating(movie["rating"])
print(result)
```

9.File Handling

File handling is used to read data from a file and write data to a file.

Basic concepts:

open() → open file
"r" → read
"w" → write
"a" → add data
read() → read file content
write() → write content
close() → close file

```python
#Write
file = open("movies.txt", "w")

file.write("Inception\n")
file.write("Interstellar\n")
file.write("Avatar\n")

file.close()

#Read
file = open("movies.txt", "r")

data = file.read()
print(data)
file.close()
```

10.Data Science Fundamental:

Data Science Fundamentals
Data Science is the process of collecting, cleaning, analysing and interpreting data to find useful information and support better decisions.

Basic Data Science Workflow

Collect Data → Clean Data → Analyse Data → Find Insights → Make Decisions

Hollywood example:

Collect → Movie name, rating, budget, collection
Clean → Remove missing/wrong data
Analyse → Compare movie ratings
Find Insight → Find the highest-rated movie
Decision → Recommend the best movies

Real-world applications:

Netflix → Movie recommendations
Amazon → Product recommendations
Banking → Fraud detection
Healthcare → Disease prediction
